# Student Performance Data Analysis Using Python
**Module 01 – Week 1 | Beginner Project**

This notebook analyzes a sample student dataset using **NumPy, Pandas and Matplotlib**. The included CSV contains fictional student records and a small number of missing marks so that data-cleaning techniques can be demonstrated.

**Learning outcomes**
- Load and inspect a CSV file.
- Find and handle missing values.
- Calculate averages and assign grades.
- Summarize subject performance and attendance.
- Visualize findings and identify students who may need support.

> The dataset is fictional and is intended only for learning and demonstration.

## 1. Import the required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
plt.rcParams["figure.figsize"] = (9, 5)
print("Libraries imported successfully.")
```

## 2. Load the student dataset

```python
# If you are running this notebook in Google Colab, upload student_performance.csv.
from google.colab import files
uploaded = files.upload()

# Read the uploaded CSV file.
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Dataset loaded successfully.")
display(df.head())
```

## 3. Explore the dataset

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nNumerical summary:")
display(df.describe())
```

## 4. Check and handle missing values

```python
print("Missing values before cleaning:")
display(df.isnull().sum())

subject_columns = ["Math", "Science", "English", "Computer"]

# Fill missing subject marks with the median of that subject.
for column in subject_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill missing attendance with its median if any attendance values are missing.
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())

print("Missing values after cleaning:")
display(df.isnull().sum())
```

## 5. Calculate average marks and grades

```python
df["Average"] = df[subject_columns].mean(axis=1)

def assign_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(assign_grade)

# Round calculated values for easier reading.
df["Average"] = df["Average"].round(2)

display(df[["Student_ID", "Name", "Average", "Grade", "Attendance"]].head(10))
```

## 6. Overall performance insights

```python
class_average = df["Average"].mean()
highest_average = df["Average"].max()
lowest_average = df["Average"].min()
attendance_average = df["Attendance"].mean()
top_student = df.loc[df["Average"].idxmax(), "Name"]

print(f"Number of students: {len(df)}")
print(f"Class average: {class_average:.2f}")
print(f"Highest student average: {highest_average:.2f}")
print(f"Lowest student average: {lowest_average:.2f}")
print(f"Average attendance: {attendance_average:.2f}%")
print(f"Top-performing student: {top_student}")
```

## 7. Subject-wise average marks

```python
subject_averages = df[subject_columns].mean().sort_values(ascending=False)
display(subject_averages.to_frame(name="Average Marks"))

plt.figure()
subject_averages.plot(kind="bar", edgecolor="black")
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks (out of 100)")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Grade distribution

```python
grade_order = ["A", "B", "C", "D", "F"]
grade_counts = df["Grade"].value_counts().reindex(grade_order, fill_value=0)
display(grade_counts.to_frame(name="Number of Students"))

plt.figure()
grade_counts.plot(kind="bar", edgecolor="black")
plt.title("Student Grade Distribution")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 9. Distribution of student averages

```python
plt.figure()
plt.hist(df["Average"], bins=8, edgecolor="black")
plt.title("Distribution of Student Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

## 10. Attendance versus academic average

```python
plt.figure()
plt.scatter(df["Attendance"], df["Average"])
plt.title("Attendance vs Student Average")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")
plt.tight_layout()
plt.show()

correlation = df["Attendance"].corr(df["Average"])
print(f"Correlation between attendance and average marks: {correlation:.2f}")
```

## 11. Identify students who may need support

```python
# These are simple project rules, not formal academic judgments.
students_needing_support = df[(df["Average"] < 60) | (df["Attendance"] < 75)]
display(students_needing_support[
    ["Student_ID", "Name", "Average", "Grade", "Attendance"]
].sort_values("Average"))

print("Students flagged for support:", len(students_needing_support))
```

## 12. Top five students

```python
top_five = df.nlargest(5, "Average")[["Student_ID", "Name", "Average", "Grade"]]
display(top_five)
```

## 13. Export the cleaned analysis

```python
output_file = "student_performance_analysis.csv"
df.to_csv(output_file, index=False)
print(f"Saved cleaned results to {output_file}")

# In Google Colab, uncomment the next two lines to download the result.
# from google.colab import files
# files.download(output_file)
```
